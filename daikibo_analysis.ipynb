```python
import pandas as pd
from google.colab import files

# 1. Load your exact file name
df = pd.read_excel('Task 5 Equality Table (1).xlsx')

# 2. Define the classification function based on the task rules
def classify_equality(score):
    if -10 <= score <= 10:
        return 'Fair'
    elif -20 <= score <= 20:
        return 'Unfair'
    else:
        return 'Highly Discriminative'

# 3. Create the 4th column "Equality class"
df['Equality class'] = df['Equality Score'].apply(classify_equality)

# 4. Preview the first 10 rows to verify
print(df.head(10))

# 5. Save and download the updated Excel file
output_name = 'Equality Table_Updated.xlsx'
df.to_excel(output_name, index=False)
files.download(output_name)
```
